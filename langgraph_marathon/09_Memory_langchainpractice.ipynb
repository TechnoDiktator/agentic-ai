# Buffer Memory

conversation buffer memory keeps a list of chat messages in a buffer and passes those into the prompt template

```python
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain.chains import LLMChain
from langchain.memory import ConversationBufferMemory
from langchain_openai import ChatOpenAI

# MessagesPlaceholder is where prior turns from memory are inserted into each prompt.
prompt = ChatPromptTemplate.from_messages(
    [
        ("system", "You are a helpful assistant. Use the conversation history when relevant."),
        MessagesPlaceholder(variable_name="chat_history"),
        ("human", "{input}"),
    ]
)

# Keep the full conversation and return it as chat messages for the placeholder.
memory = ConversationBufferMemory(
    memory_key="chat_history",
    return_messages=True,
)

# LLMChain loads chat_history from memory before each call and saves the new turn afterward.
conversation = LLMChain(
    llm=ChatOpenAI(model="gpt-4.1-mini"),
    prompt=prompt,
    memory=memory,
)

converstaion = LLMChain(
    llm=llm,
    prompt=prompt,
    verbose=True,
    memory=memory

)
```
