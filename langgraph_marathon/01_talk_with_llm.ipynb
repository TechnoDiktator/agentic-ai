```python
# Chat model

from langchain_openai import ChatOpenAI
llmModel = ChatOpenAI(model="gpt-4.1-mini")
```

<!-- llmModel -->

```python
from langchain_core.messages import HumanMessage, SystemMessage

"""
A system message sets the assistant’s role and guidelines; a human message is the user’s request.

In your example:

System: “You are an expert historian specializing in the Kennedy family.”
Human: “Tell me one curious thing about JFK.”
The system message guides how the model should answer. The human message says what to answer.

"""


messages = [
    SystemMessage(content="You are an expert historian specializing in the Kennedy family."),
    HumanMessage(content="Tell me one curious thing about JFK."),
]

response = llmModel.invoke(messages)
print(response.content)
```

```python
## Alternative llm


# here we will use groq


import os 
from dotenv import load_dotenv , find_dotenv
_ = load_dotenv(find_dotenv())  # read local .env file
```

```python
from langchain_groq import ChatGroq

groq20bChatModel = ChatGroq(model="openai/gpt-oss-20b", api_key=os.environ.get("GROQ_API_KEY"))
```

```python
groq120bChatModel = ChatGroq(model="openai/gpt-oss-120b", api_key=os.environ.get("GROQ_API_KEY"))
```

```python
messages = [
    ("system", "You are an expert historian specializing in the Kennedy family."),
    ("human", "Tell me one curious thing about JFK."),
]

groq20bResponse = groq20bChatModel.invoke(messages)
groq120bResponse = groq120bChatModel.invoke(messages)


print(groq20bResponse)
print(groq120bResponse)
```
