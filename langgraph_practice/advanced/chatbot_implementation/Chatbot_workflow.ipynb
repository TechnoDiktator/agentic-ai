```python
from langgraph.graph import StateGraph, START, END
from typing import TypedDict, Annotated
from langchain_core.messages import BaseMessage, HumanMessage
from langchain_openai import ChatOpenAI
from dotenv import load_dotenv
```

```python
from langgraph.checkpoint.memory import MemorySaver
```

```python
load_dotenv()
```

```python
llm = ChatOpenAI()
```

```python
from langgraph.graph.message import add_messages

class ChatState(TypedDict):

    messages: Annotated[list[BaseMessage], add_messages]
```

```python
def chat_node(state: ChatState):
    # The message reducer supplies the conversation history accumulated so far.
    messages = state['messages']

    # Send the full conversation, not only the newest user message, to the LLM.
    response = llm.invoke(messages)

    # add_messages appends this assistant response to the checkpointed history.
    return {'messages': [response]}
```

```python
# MemorySaver stores graph checkpoints in memory, separated by thread_id.
# History lasts while this kernel/process is running; it is not durable storage.
checkpoint = MemorySaver()

graph = StateGraph(ChatState)

# Each call runs this node, which reads and extends the messages state.
graph.add_node('chat_node', chat_node)

graph.add_edge(START, 'chat_node')
graph.add_edge('chat_node', END)

# Attach the in-memory checkpointer so invokes can resume the same thread.
chatbot = graph.compile(checkpointer=checkpoint)
```

```python
chatbot
```

```python
thread_id = "2"
initial_state = {
    'messages': [HumanMessage(content='What is my name?')]
}

config = {'configurable': {'thread_id': thread_id}}
response = chatbot.invoke(initial_state, config=config)
```

## Conversation history and persistence
`ChatState.messages` uses the `add_messages` reducer, and `MemorySaver` checkpoints that state by `thread_id`. Calls using the same `thread_id` continue the same conversation while this Python kernel/process remains alive; they do not start fresh on every input. In the interactive loop below, the ID stays `"1"` for every turn. The earlier one-off example uses `"2"`, so it is a separate conversation.

`MemorySaver` is in-memory only. Restarting the kernel or process clears its checkpoints. For history that survives restarts, use a persistent checkpointer such as SQLite or Postgres.


```python
# Reuse one thread ID so each input continues the same in-memory conversation.
thread_id = "1"

while True:
    user_message = input('Type here: ')

    print('User:', user_message)

    if user_message.strip().lower() in ['exit', 'quit', 'bye']:
        break

    # The same config tells the checkpointer which conversation history to resume.
    config = {'configurable': {'thread_id': thread_id}}
    response = chatbot.invoke({'messages': [HumanMessage(content=user_message)]}, config=config)
    print('AI:', response['messages'][-1].content)
```
