# Retrieval-Augmented Generation (RAG)

RAG combines retrieved information with a language model's generation. Instead of relying only on what the model learned during training, the application finds relevant external content and supplies it as context for the current question.

## Typical Flow

1. **Load** source material, such as a file or web page.
2. **Split and index** larger collections into chunks, often representing them as embeddings in a vector store.
3. **Retrieve** the chunks most relevant to the user's question.
4. **Augment** the prompt with both the question and retrieved context.
5. **Generate** an answer grounded in that context.

## In This Notebook

`WikipediaLoader` fetches one Tesla article. The prompt cell includes its text in `{context}` alongside the question, and `chatModel` generates the response. This demonstrates loading and context augmentation; it passes the whole article directly rather than searching a chunked index. A fuller RAG pipeline typically chunks and indexes documents, then retrieves only relevant passages for each question.

RAG supplies context at request time; it does not retrain or fine-tune the model. The answer can still be incomplete or incorrect, so keep the retrieved source available when checking important claims.

````python
from langchain_community.document_loaders import TextLoader

loader = TextLoader("data/be-good.txt")
loaded_data = loader.load()
loaded_data
````

````python
from langchain_text_splitters import RecursiveCharacterTextSplitter

# Split loaded Documents into overlapping chunks for downstream retrieval.
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=50,
    add_start_index=True,
)
chunks = text_splitter.split_documents(loaded_data)

print(f"Source documents: {len(loaded_data)}")
print(f"Text chunks: {len(chunks)}")
print("First chunk:")
print(chunks[0].page_content)
print("Metadata:", chunks[0].metadata)
````

### RecursiveCharacterTextSplitter

`RecursiveCharacterTextSplitter` breaks long text into smaller pieces while trying to keep natural boundaries. By default, it tries paragraph breaks (`\n\n`), then line breaks (`\n`), spaces, and finally individual characters. This fallback helps it handle text that has no convenient paragraph boundaries.

In the example above:

- `chunk_size=500` sets the target chunk length. The default length function counts characters.
- `chunk_overlap=50` repeats some text between neighboring chunks so context near a boundary is less likely to be lost.
- `split_documents(loaded_data)` splits LangChain `Document` objects and carries their metadata into each chunk.
- `add_start_index=True` adds each chunk's starting character offset to its metadata.

Chunk size and overlap are tuning choices: smaller chunks can make retrieval more focused, while larger chunks retain more context. Compared with the `CharacterTextSplitter` below, the recursive splitter can fall back through several separator types.

````python
from langchain_text_splitters import CharacterTextSplitter

# Split on line boundaries; unlike the recursive splitter, this uses one separator.
character_splitter = CharacterTextSplitter(
    separator="\n",
    chunk_size=500,
    chunk_overlap=50,
    length_function=len,
)
character_chunks = character_splitter.split_documents(loaded_data)

print(f"Character-split chunks: {len(character_chunks)}")
print(f"First chunk length: {len(character_chunks[0].page_content)} characters")
print(character_chunks[0].page_content)
print("Metadata:", character_chunks[0].metadata)
````

# Chroma Vector Store

Chroma is a vector store (often called a vector database) that keeps document chunks together with their embeddings and metadata. Embeddings represent text as numeric vectors, allowing semantically similar text to be found even when it does not use the exact same words as a query.

In the code cell below, `OpenAIEmbeddings` creates a vector for each chunk, and `Chroma.from_documents(...)` stores those vectors with the source text and metadata. Embedding creation sends text to the OpenAI API and may incur usage charges.

## Retrieve with a Retriever

A **retriever** is a standard interface for finding documents relevant to a text query. `vector_db.as_retriever(...)` adapts the Chroma store into a retriever that can be invoked directly or composed into an LCEL chain. Here, `search_kwargs={"k": 2}` asks for the two closest chunks:

```python
retriever = vector_db.as_retriever(
    search_type="similarity",
    search_kwargs={"k": 2},
)
results = retriever.invoke("What does the speech say about healthcare?")
```

`similarity_search(query, k=2)` performs the same basic lookup directly on the vector store; the retriever is more reusable in chains. Retrieval returns `Document` objects containing `page_content` and metadata.

The current `Chroma.from_documents(...)` call creates a temporary in-memory store. To keep it on disk between runs, pass a `persist_directory`, for example `persist_directory="./chroma_db"`, when creating the store. Reuse the same embedding configuration when reopening it.

````python
from langchain_community.document_loaders import TextLoader
from langchain_openai import OpenAIEmbeddings
from langchain_text_splitters import CharacterTextSplitter
from langchain_chroma import Chroma

loaded_document = TextLoader(
    "data/state_of_the_union.txt",
    encoding="utf-8",
).load()

text_splitter = CharacterTextSplitter(
    separator="\n",
    chunk_size=500,
    chunk_overlap=50,
    length_function=len,
)

chunk_of_text = text_splitter.split_documents(loaded_document)

vector_db = Chroma.from_documents(
    documents=chunk_of_text,
    embedding=OpenAIEmbeddings(),
)

print(f"Loaded {len(loaded_document)} document(s) into {len(chunk_of_text)} chunks.")
print("Vector store created successfully.")
````

````python
len(chunk_of_text)  
````

````python
vector_db = Chroma.from_documents(
    documents=chunk_of_text,
    embedding=OpenAIEmbeddings(),
)
````

````python
question = "What did he sayabout ketangi brown jackson?"

retriever = vector_db.as_retriever(
    search_type="similarity",
    search_kwargs={"k": 2},
)
retrieved_documents = retriever.invoke(question)

print(f"Retrieved {len(retrieved_documents)} document chunks:")
for index, document in enumerate(retrieved_documents, start=1):
    print(f"\nChunk {index}:")
    print(document.page_content)
    print("Metadata:", document.metadata)
````

````python
from uuid import uuid4
from langchain_openai import OpenAIEmbeddings
from langchain_chroma import Chroma

sample_text = "Chroma stores text alongside vector embeddings for similarity search."
embedding_model = OpenAIEmbeddings()

demo_vector_db = Chroma.from_texts(
    texts=[sample_text],
    embedding=embedding_model,
    collection_name=f"embedding_demo_{uuid4().hex[:8]}",
)

stored_data = demo_vector_db.get(include=["documents", "embeddings"])
stored_text = stored_data["documents"][0]
stored_embedding = stored_data["embeddings"][0]

print("Stored text:", stored_text)
print("Embedding dimensions:", len(stored_embedding))
print("First 10 embedding values:", list(stored_embedding[:10]))
````
